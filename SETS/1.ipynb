```python

set1 = set()
print(set1)

set1.add(4)
set1.add('STUDY')
set1.add((6,7))
print(set1)

for i in range(5):
    set1.add(i)
print(set1)

set2 = set([4,5,(6,7)])
set2.update([10,11])
print(set2)
```

```python
set3 = set(["WE","HAVE","TO","STUDY","HARD"])
print(set3)

for i in set3:
    print(i, end = " ")
    
print("\n")
    
# CHECKING IN ELEMENT USING KEYBOARD
print("yes" in set3) # returns true or false
    
```

```python
#DELETION OF ELEMENTS
set1 = set([1,2,3,3,5,6,7,4,3,4,2,5,6,7,3,7])
print(set1)
set1.remove(4)
set1.remove(3)
print(set1)
set1.discard(5)
print(set1)
```
