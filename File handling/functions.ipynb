```python
# READING A FILE
file = open('1.txt')
for i in file:
    print(i)
```

```python
# COUNTING LINES IN A FILE
file = open('1.txt')
count = 0
for i in file:
    count = count + 1

print("Line count: ", count)
```

```python
# READING THE WHOLE FILE
file = open('1.txt')
read = file.read()
print(len(read))
print(read[:175])
```

```python
# SEARCHING THROUGH FILE
file = open('1.txt')
for i in file:
    if i.startswith('IS'):
        print(i)
        
```

```python
# SEARCHING THROUGH FILE(FIXED)
file = open('1.txt')
for i in file:
    i = i.rstrip()             # the line will strio on output
    if i.startswith('IS'):
        print(i)
```

```python
# SKIPPING WITH CONTINUE
file = open('1.txt')
for i in file:
    i = i.rstrip()
    if not i.startswith('IS'):
        continue
    print(i)
```

```python
# USING in TO SELECT LINES BY WORDS
file = open('1.txt')
for i in file:
    i = i.rstrip()
    if not 'NO' in i:
        continue
    print(i)        
```
