```python
#============ REGEX =============
import re

# USING find() FUNCTION
file = open('1.txt')
for i in file:
    i = i.rstrip()
    if i.find('IS') >= 0:
        print(i)
```

```python
#============ REGEX =============
import re

# USING re.search() FUNCTION
file = open('1.txt')
for i in file:
    i = i.rstrip()
    if re.search('IS', i):
        print(i)
        
```

```python
# MATCHING AND EXTRACTING DATA
# re.findall()
import re
x = 'My 2 favourite numbers are 19 and 49'
y = re.findall('[0-9]+', x)
z = re.findall('[0-8]+', x)
a = re.findall('[aeiou]+', x)
b = re.findall('[aeiou]', x)

print(y)
print(z)
print(a)
print(b)
```

```python
#      GREEDY MATCHING   VS     NON GREEDY MATCHING
import re
x = 'From: Using the : character'
y = re.findall('^F.+?:', x)         # NON GREEDY MATCHING
z = re.findall('^F.+:', x)    # GREEDY MATCHINF BECAUSE IT GOES TO NEXT COLON
a = re.findall('^F.+e', x)   # GREEDY MATCHING
b = re.findall('^F.+?e', x)  # NON GREEDY MATCHING
c = re.findall('U.+?e', x)   # NON GREEDY
d = re.findall('c.+r', x)    # GREEDY
e = re.findall('th.+?er', x)   # NON GREEDY

f = re.findall('From.+?acter', x)  # NON GREEDY
g = re.findall('From.+r', x)      # GREEDY

print(y)
print(z)
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
```

```python
#     FINE TUNING STRING EXTRACTION
x = 'FROM stephen.marquard@uct.ac.za Sat Jan 09:14:16 2008'
y = re.findall('\\S+@\\S+', x)
z = re.findall('^From (\\S + @ \\S Jan)', x)
print(y)
print(z)
```
