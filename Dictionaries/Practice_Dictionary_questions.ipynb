```python
marks = {"Math": 85,"Physics": 78,"Programming": 92,"English": 70}
sum = 0
for i in marks.values():
    sum = sum + i
    
print("Total sum is:", sum)
```

```python
marks = {"Math": 85,"Physics": 78,"Programming": 92,"English": 70}
max = 0
for i in marks.values():
    if i > max:
        max = i
        
print("Max number is: ", max)
```

```python
marks = {"Math": 85,"Physics": 78,"Programming": 92,"English": 70}
print("After increment 5 marks in each subject:")
for i in marks:
    marks[i] = marks[i] + 5
    
print(marks)
```
