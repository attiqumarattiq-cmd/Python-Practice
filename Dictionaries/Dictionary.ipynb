```python
# NESTED DICTIONARY
Dict1 = {1: 'FAST', 2: 'NU', 3: {'A': 'Welcome', 'B': 'You', 'C': 'All'}}
print(Dict1)
```

```python
Dict = {}
print("Empty Dictionary: ", Dict)
# Adding elements one at a time
Dict[1] = 'Education'
Dict[2] = 'For'
Dict[3] = 1
print("Dictionary after adding 3 elements: ", Dict)
# Adding set of values to a single key
Dict['Value_set'] = 2,3,4
print(Dict)
# UPDATING existing key's value
Dict[2] = 'Welcome'
print('After updating the key value: ', Dict)
```

```python
# ACCESSSING A ELEMENT USING KEY
Dict = {1: 'Education', 2: 'For', 3: 'All', 'yes': 'You', 4 : { 1: 'Yes', 2: 'I', 3: 'am', 'talking': 'to you'}}
print('After accessing dictionary using key: ')
print(Dict[1])
print(Dict['yes'])
print(Dict[4])
```

```python
#ACCESSING ELEMENT USING GET()
Dict = {1: 'Education', 2: 'For', 3: 'All', 'yes': 'You', 4 : { 1: 'Yes', 2: 'I', 3: 'am', 'talking': 'to you'}}
print('Accessing element using get(): ')
print(Dict.get(1))
print(Dict.get('yes'))
print(Dict.get(4))
```

```python
# Accessing element of NESTED DICTIONARY
Dict = {1: {1: 'YEAH', 2: 'BRO', 3: 'This'}, 2: {1: 'is', 2: {1: 'me', 2: 'Why?'}}}
print("Accessing Nested Dictionary: ")
print(Dict[1])
print(Dict[1][2])
print(Dict[2][2][1])
print(Dict[2][2][2])
```

```python
# ====== DICTIONARY METHODS =============
dict1 = {1: 'Python', 2: 'C++', 3: 'Java', 4: 'HTML'}
# copy() method
dict2 = dict1.copy()
print("1. After copying dictionary: ")
print(dict2)
# get() method
print('2. After using get() function for specific part: ')
print(dict1.get(2))
# items() method
print("3. After printing items of Dictionary using items(): ")
print(dict1.items())
# keys() method
print('4. After obtaining the keys using keys(): ')
print(dict1.keys())
# pop() method
print("5. Applying pop(): ")
print(dict1.pop(2))
# popitem() method
print("6. Using popitem(): ")
print(dict1.popitem())
# update() method
print("7. Using update() function to update the values: ")
dict1.update({2: 'Assembly language'})
print(dict1)
# Values() method
print("8. Shows the values of Dictionaries using Values() function: ")
print(dict1.values())
# clear() method
dict1.clear()
print('9. After clearing dictionary: ')
print(dict1)
```
